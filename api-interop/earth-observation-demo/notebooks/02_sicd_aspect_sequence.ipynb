ทดลองเฟรมย่อยด้วย deskew และ spectral windows บน full cross-range support ของ narrow range strip รอบศูนย์กลาง SICD เฟรมทับซ้อนกัน ไม่ใส่เวลาเฟรมหรือความเร็วรถที่ยังไม่ได้ยืนยัน

```python
from pathlib import Path
import sys, os
ROOT = Path(os.environ.get('EO_DEMO_ROOT', Path.cwd())).resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
assert (ROOT / 'eo_engine').exists(), 'Set EO_DEMO_ROOT to the Earth Observation Demo source directory (local checkout or mounted Drive).'
sys.path.insert(0, str(ROOT))
```

```python
from eo_engine.catalog import scenes, scene_dir
from eo_engine.sequence import sequence
from IPython.display import display, Image
import json
row = next(r for r in scenes() if r['site_id']=='suvarnabhumi')
folder = scene_dir(row['id'])
```

```python
result = json.loads((folder/'sequence.json').read_text()) if (folder/'sequence.json').exists() else sequence(row['id'], frames=8, progress=lambda message,pct: print(pct,message))
print(result['warning'])
print('Bytes read:', result['downloaded_bytes'])
display(Image(filename=str(folder/'sequence.gif')))
```

```python
for frame in result['frames']:
    print(frame['index'], frame['spectral_start'], frame['spectral_stop'], 'UTC:', frame['time_utc'])
```
